# API LLMs

Available Protolink Native LLM wrappers

- OpenAI API: OpenAILLM - Set `OPENAI_API_KEY` env variable or pass it to the constructor
- Anthropic API: AnthropicLLM - Set `ANTHROPIC_API_KEY` env variable or pass it to the constructor
- Gemini API: GeminiLLM - Set `GEMINI_API_KEY` env variable or pass it to the constructor
- Deepseek API: DeepseekLLM - Set `DEEPSEEK_API_KEY` env variable or pass it to the constructor
- Hugging Face Hub API: HuggingFaceLLM - Set `HF_API_TOKEN` env variable or pass it to the constructor - **TBD**


Parametes you can pass include:
- `api_key`: API key for the provider (alternative to env variables)
- `model`: The model to use (default varies by provider)
- `model_params`: Dictionary of model-specific parameters
- `base_url`: Base URL for the provider API

#### OpenAI API LLM

```python
from protolink.llms.api import OpenAILLM

llm = OpenAILLM()
```

#### Anthropic API LLM

```python
from protolink.llms.api import AnthropicLLM

llm = AnthropicLLM()
```

#### Gemini API LLM

```python
from protolink.llms.api import GeminiLLM

llm = GeminiLLM(api_key="<YOUR_API_KEY>")
```

#### DeepSeek API LLM

```python
from protolink.llms.api import DeepSeekLLM

llm = DeepSeekLLM()
```

#### Ollama Server LLM

You can use the Ollama Server LLM by setting the `OLLAMA_HOST` environment variable to the URL of your Ollama server or pass it directly to the `base_url` parameter.
You should have the Ollama server running and accessible at the specified URL, otherwise the LLM will not work.

```python
from protolink.llms.server import OllamaLLM

llm = OllamaLLM(base_url="http://localhost:11434")
```

### Set Text Input question

```python
# Input
query = "What's the capital of Greece?"
```

### Test normal response

By using the `system prompt` provided by protolink, you'll notice that the response is returned as a dict where the answer is of type `text`

```python
print("\nTesting non-streaming response:")
response = llm.chat(query)
print(f"Response:{response}")
```

### Test streaming response

```python
async def test_streaming():
    async for chunk in llm.chat(query, streaming=True):
        print(chunk, end="", flush=True)
    print("\n")


await test_streaming()
```
